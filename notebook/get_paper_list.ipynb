```python
import requests
import json
import pandas as pd

term = "drug AND (\"toxicity prediction\" OR \"predicting toxicity\") AND \"graph neural network\""
sort = "RELEVANT_DESC"
limit = "50"
searchDateFrom = "2015-01-01T00:00:00.000Z"
skip = 0
categoryIds = "605c72ef153207001f6470ce"

allhits = pd.DataFrame([], columns=["title", "abstract", "published date", "doi"])

while True:
    url = "https://chemrxiv.org/engage/chemrxiv/public-api/v1/items?term=" + term + "&sort=" + sort + "&limit=" + limit + "&searchDateFrom=" + searchDateFrom +"&skip=" + f"{skip}" + "&categoryIds=" + categoryIds
    
    response = requests.get(url)
    
    if response.status_code != 200:
        print("ERROR!")
        break
    else:
        data = response.json()
        if len(data["itemHits"]) == 0:
            print("FINISHED!")
            break
        else:
            for i, hit in enumerate(data["itemHits"]):
                title = hit["item"]["title"]
                abstract = hit["item"]["abstract"]
                date = hit["item"]["publishedDate"]
                doi = hit["item"]["doi"]
                allhits = pd.concat([allhits, pd.DataFrame(data=[[title, abstract, date, doi]], columns=["title", "abstract", "published date", "doi"])], axis=0)
            skip += 50
            print(f"No.{skip} finished")
```

```python
allhits.reset_index(inplace=True, drop=True)
allhits.to_csv("/workspace/liz/SRToxPred/survey_list/240701_ChemRxiv_hits.csv")
```

# Get Arxiv paper

```python
import requests
import arxiv
from datetime import datetime, timedelta
import pytz
import numpy as np
import pandas as pd
```

```python
client = arxiv.Client()
end = datetime.now(pytz.timezone("Asia/Tokyo"))
start = "201501010000"
end = end.strftime("%Y%m%d") + "0000"

res_list = []

res = client.results(arxiv.Search(query=f'%28%28drug OR chemical%29 AND toxicity AND %28%28neural network%29 OR %28deep learning%29%29 AND %28graph OR SMILES OR strings%29%29 AND submittedDate:[{start} TO {end}]'))
cnt = 0
for r in res:
    info = []
    info.append(r.title)
    info.append(r.summary)
    info.append(str(r))
    res_list.append(info)
    cnt += 1

res_df = pd.DataFrame(columns=["Title", "Abstract", "Link"], index=np.arange(len(res_list)))

if cnt == 0 and len(res_list) == 0:
    print("やり直せｋｓ")
else:
    print(f"とりあえずおｋ、件数は{len(res_list)}")
    for i in range(cnt):
        res_df.loc[i] = [res_list[i][0], res_list[i][1], res_list[i][2]]
    display(res_df.head())
    print(res_df.shape)
    res_df.to_csv("/workspace/liz/SRToxPred/survey_list/Arxiv_list_240818.csv")
```
